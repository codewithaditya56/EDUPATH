```python
import sklearn
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
from sklearn.datasets import make_moons
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
X, y = make_moons(n_samples=100, noise=0.15)
polynomial_svm_clf = Pipeline([
 ("poly_features", PolynomialFeatures(degree=3)),
 ("scaler", StandardScaler()),
 ("svm_clf", LinearSVC(C=10, loss="hinge"))
 ])
polynomial_svm_clf.fit(X, y)
```

###

```python
from sklearn.svm import SVC
poly_kernel_svm_clf = Pipeline([
 ("scaler", StandardScaler()),
 ("svm_clf", SVC(kernel="poly", degree=3, coef0=1, C=5))
 ])
poly_kernel_svm_clf.fit(X, y)
```

```python
rbf_kernel_svm_clf = Pipeline([
 ("scaler", StandardScaler()),
 ("svm_clf", SVC(kernel="rbf", gamma=5, C=0.001))
 ])
rbf_kernel_svm_clf.fit(X, y)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_boundary(clf, X, y, ax, title):
    # Create grid
    x0s = np.linspace(-1.5, 2.5, 200)
    x1s = np.linspace(-1, 2, 200)
    x0, x1 = np.meshgrid(x0s, x1s)
    X_grid = np.c_[x0.ravel(), x1.ravel()]
    y_pred = clf.predict(X_grid).reshape(x0.shape)

    # Plot decision regions
    ax.contourf(x0, x1, y_pred, alpha=0.3, cmap=plt.cm.brg)
    ax.scatter(X[:, 0], X[:, 1], c=y, s=30, cmap=plt.cm.brg, edgecolors='k')
    ax.set_title(title)

# Create figure with 3 subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

plot_boundary(polynomial_svm_clf, X, y, axes[0], "Poly Features + Linear SVM")
plot_boundary(poly_kernel_svm_clf, X, y, axes[1], "Polynomial Kernel SVM")
plot_boundary(rbf_kernel_svm_clf, X, y, axes[2], "RBF Kernel SVM")

plt.show()
```
